```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model = "gemma2",
    temperature = 0.8,
    num_predict = 256,
    # other params ...
)
```

```python
messages = [
    ("system", "You are an experienced narrator of fantasy stories and of role playing games such as Dungeons and Dragons, Diablo and The Legend of Zelda. Your audience is a group of people who are interested in fantasy stories. Your goal is to be the Dungeon Master in a Dungeons and Dragons campaign, generate a story that is engaging and entertaining."),
    ("human", "Let's begin the story in the setup of the campaign. The adventurers are in Kakariko Village, a small town at the foot of Death Mountain. The village is known for its windmill, which is powered by the wind blowing down from the mountain. The adventurers are in the village square, where they see a group of villagers gathered around a well. The villagers are talking excitedly and pointing down into the well. The adventurers approach the well and look down into it. They see a dark tunnel leading down into the earth. The villagers tell the adventurers that the tunnel leads to the lair of a shadow that has been terrorizing the village. The villagers ask the adventurers to go down into the tunnel and defeat the shadow monster. The adventurers agree and prepare to descend into the tunnel."),
]
response = llm.invoke(messages)

print(response.content)
```
